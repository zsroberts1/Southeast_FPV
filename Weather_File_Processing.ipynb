```python
import os

i = 0

source_path = "/content/drive/MyDrive/DATA/Weather_data"
destination_path = "/content/drive/MyDrive/DATA/Weather_data/New_weather_files"
extension = ".csv"

# Ensure the destination folder exists
os.makedirs(destination_path, exist_ok=True)

for x in os.listdir(source_path):
    if x.endswith(".csv"):
        old_name = os.path.join(source_path, x)

        # Extract new file name
        file_split = os.path.splitext(os.path.split(old_name)[1])[0].rsplit("_", 5)
        ExtractName = file_split[1] + "_" + file_split[2]
        fullName = ExtractName + extension
        new_name = os.path.join(destination_path, fullName)

        # Check if the file already exists
        if os.path.exists(new_name):
            print(f"File already exists: {new_name}. Skipping {old_name}.")
            continue  # Skip the current file

        # Perform the renaming
        print(f"Renaming: {old_name} -> {new_name}")
        os.rename(old_name, new_name)
        i += 1

print(f"Total files renamed: {i}")
```
